```python
from ultralytics import YOLO

# Adapter ce chemin vers ton dossier dans Google Drive
# Ex: si ton projet est dans Drive/first-YOLO-project/
PROJECT_PATH = '/content/drive/MyDrive/first-YOLO-project'

model = YOLO('yolo11n.pt')

model.train(
    data=f'{PROJECT_PATH}/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,            # GPU (0 = premier GPU)
    workers=2,
    patience=10,         # Early stopping
    project=f'{PROJECT_PATH}/runs',
    name='coca_pepsi'
)
```

```python
# Evaluer le modele
results = model.val()
print(f"mAP50: {results.box.map50:.4f}")
print(f"mAP50-95: {results.box.map:.4f}")
```
